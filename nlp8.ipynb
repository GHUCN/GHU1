```python
from sklearn.decomposition import LatentDirichletAllocation
from sklearn.feature_extraction.text import CountVectorizer
```

```python
docs = [
    "he studies in GECA",
    "GECA is an engineering college",
    "Politics and government",
    "Elections and democracy",
    "Artificial intelligence and machine learning",
    "Deep learning for image processing",
    "Cricket and football are popular sports",
    "Health and fitness are important",
    "Python programming for data science",
    "Cloud computing and cybersecurity"
]

# Convert text into matrix
vectorizer = CountVectorizer(stop_words='english')

X = vectorizer.fit_transform(docs)

# Create LDA Model with 10 Topics
lda = LatentDirichletAllocation(
    n_components=10,
    random_state=42
)

# Train model
lda.fit(X)

# Get feature names
terms = vectorizer.get_feature_names_out()

# Display Topics
for topic_index, topic in enumerate(lda.components_):

    print(f"\nTopic {topic_index + 1}:")

    top_words = [terms[i] for i in topic.argsort()[-5:]]

    print(top_words)
```

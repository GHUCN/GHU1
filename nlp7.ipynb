```python
import random

# Expanded lists of positive and negative words/phrases
positive_words = [
    "excellent", "fantastic", "amazing", "wonderful", "great", "superb",
    "love it", "highly recommend", "perfect", "brilliant", "outstanding",
    "delightful", "impressive", "top-notch", "superior", "joyful",
    "efficient", "effective", "user-friendly", "seamless", "smooth",
    "thrilled", "satisfied", "pleased", "happy", "enjoyable", "convenient",
    "reliable", "sturdy", "innovative", "insightful", "captivating",
    "well-made", "robust", "powerful", "responsive", "intuitive", "easy to use"
]

negative_words = [
    "terrible", "awful", "horrible", "worst", "disappointing", "bad",
    "waste of money", "not happy", "broke quickly", "poor design", "difficult",
    "frustrating", "useless", "regret buying", "scam", "ripoff",
    "unacceptable", "mediocre", "defective", "flawed", "issues",
    "glitches", "unresponsive", "buggy", "crashes", "slow", "clunky",
    "misleading", "overpriced", "underperforming", "fragile", "pure junk",
    "disgruntled", "annoying", "problematic", "subpar", "below average"
]

neutral_words = [
    "it's okay", "average", "acceptable", "not bad", "indifferent",
    "standard", "functional", "basic", "common", "regular", "expected",
    "decent", "adequate", "neither good nor bad", "middle-of-the-road"
]

def generate_synthetic_review(sentiment_type, num_sentences=2):
    review_sentences = []
    for _ in range(num_sentences):
        if sentiment_type == 'positive':
            words = random.sample(positive_words, min(len(positive_words), random.randint(3, 7))) # Ensure no more than available words
            template = random.choice([
                "This product is {}", "I find this to be {}", "The service was quite {}",
                "Overall, a very {} experience", "Absolutely {}!", "I am {} with this."
            ])
        elif sentiment_type == 'negative':
            words = random.sample(negative_words, min(len(negative_words), random.randint(3, 7))) # Ensure no more than available words
            template = random.choice([
                "This product is {}", "I find this to be {}", "The service was quite {}",
                "Overall, a very {} experience", "Absolutely {}!", "I am {} with this."
            ])
        else:
            words = random.sample(neutral_words, min(len(neutral_words), random.randint(2, 5))) # Ensure no more than available words
            template = random.choice([
                "It's {}", "The quality is {}", "My experience was {}",
                "This is rather {}", "I found it to be {}."
            ])

        review = template.format(', '.join(words)) + '.'
        review_sentences.append(review)
    return ' '.join(review_sentences)

# Generate a larger synthetic dataset
num_samples_per_sentiment = 500  # Generate 500 positive and 500 negative reviews

new_texts = []
new_labels = []

for _ in range(num_samples_per_sentiment):
    new_texts.append(generate_synthetic_review('positive'))
    new_labels.append(1) # Positive

    new_texts.append(generate_synthetic_review('negative'))
    new_labels.append(0) # Negative

# Shuffle the dataset to mix positive and negative reviews
combined = list(zip(new_texts, new_labels))
random.shuffle(combined)
new_texts_shuffled, new_labels_shuffled = zip(*combined)

print(f"Generated {len(new_texts_shuffled)} synthetic reviews.")
print("First 5 synthetic reviews:")
for i in range(min(5, len(new_texts_shuffled))):
    print(f"- '{new_texts_shuffled[i]}' -> {new_labels_shuffled[i]}")

texts = list(new_texts_shuffled)
labels = list(new_labels_shuffled)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report

# Define a mapping for definite answers
label_map = {0: 'Negative', 1: 'Positive'}

# Split data into training and testing sets
X_train_text, X_test_text, y_train, y_test = train_test_split(texts, labels, test_size=0.3, random_state=42)

# Vectorize the text data using TF-IDF, including bigrams
vectorizer = TfidfVectorizer(ngram_range=(1, 2))
X_train = vectorizer.fit_transform(X_train_text)
X_test = vectorizer.transform(X_test_text)

# Train Multinomial Naive Bayes classifier
model = MultinomialNB()
model.fit(X_train, y_train)

# Make predictions on the test set
y_pred = model.predict(X_test)

# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
report = classification_report(y_test, y_pred, target_names=[label_map[0], label_map[1]])

print(f"Model Accuracy: {accuracy:.2f}")
print("Classification Report:\n", report)

# Demonstrate prediction on new data (same as before)
new_texts_for_prediction = [
    "This is a great item!",
    "I absolutely despise this.",
    "It's okay, not good not bad.",
    "What a phenomenal experience, truly amazing!",
    "This product broke on the first use, utterly disappointing.",
    "Absolutely love it, exceeds expectations!",
    "Never again, completely dissatisfied.",
    "Highly effective and user-friendly."
]

X_new = vectorizer.transform(new_texts_for_prediction)
new_predictions_numeric = model.predict(X_new)
new_predictions_definite = [label_map[pred] for pred in new_predictions_numeric]

print("\nPredictions for new texts:")
for i, text in enumerate(new_texts_for_prediction):
    print(f"- '{text}' -> {new_predictions_definite[i]}")
```

```python
import random
import nltk
from nltk.util import ngrams
```

```python
  nltk.download('punkt')
  nltk.download('punkt_tab')
```

```python
text = "Learning Ai was a better expereince than ever, This has been a great journey "
tokens = nltk.word_tokenize(text)

# Generate n-grams
unigrams = list(ngrams(tokens, 1))
bigrams = list(ngrams(tokens, 2))
trigrams = list(ngrams(tokens, 3))

print("Bigrams:", bigrams)
print("Trigrams:", trigrams)

# Simple text generation (bigram model)
def generate_text(bigrams, start_word, length=10):
    current = start_word
    result = [current]

    for _ in range(length):
        candidates = [b[1] for b in bigrams if b[0] == current]
        if not candidates:
            break
        current = random.choice(candidates)
        result.append(current)

    return ' '.join(result)

print("Generated Text:", generate_text(bigrams, "I"))
```

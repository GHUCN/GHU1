```python
import nltk
import heapq
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab')
```

```python
text = """
Artificial Intelligence is rapidly transforming industries around the globe.
It allows machines to learn from large amounts of data and make intelligent decisions.
AI is widely applied in healthcare, finance, education, and other sectors.
At the same time, it introduces challenges related to privacy, ethics, and job displacement.
Many organizations are investing heavily in AI research and development.
"""

stop_words = set(stopwords.words("english"))
words = word_tokenize(text.lower())
freq_table = {}
for word in words:
    if word not in stop_words:
        freq_table[word] = freq_table.get(word, 0) + 1
sentences = sent_tokenize(text)
sentence_scores = {}
for sent in sentences:
    for word in word_tokenize(sent.lower()):
        if word in freq_table:
            sentence_scores[sent] = sentence_scores.get(sent, 0) + freq_table[word]

summary_sentences = heapq.nlargest(2, sentence_scores, key=sentence_scores.get)
extractive_summary = " ".join(summary_sentences)
print("Extractive Summary:")
print(extractive_summary)
```

```python
from summa.summarizer import summarize

textrank_summary = summarize(text, ratio=0.5)
print("\nExtractive Summary (TextRank):")
print(textrank_summary)
```

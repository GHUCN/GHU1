```python
from gensim.models import Word2Vec
import numpy as np
```

```python
sentences = [["event", "was", "sensational"], ["computer vision", "is", "incredible"]]

# Word2Vec
model = Word2Vec(sentences, vector_size=100, window=5, min_count=1)
print("Word2Vec vector for 'event':", model.wv['event'])

# Load GloVe
def load_glove(file):
    embeddings = {}
    with open(file, encoding='utf8') as f:
        for line in f:
            values = line.split()
            word = values[0]
            vector = np.array(values[1:], dtype='float32')
            embeddings[word] = vector
    return embeddings
```
